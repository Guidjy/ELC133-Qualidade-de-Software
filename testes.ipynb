## Configuração da biblioteca

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
# define chrome como navegador
chrome_options = Options()

# executa chrome
driver = webdriver.Chrome(options=chrome_options)

# configura o webdriver para esperar no máximo dez segundos
wait = WebDriverWait(driver, 10)
```

```python
# vai para a página da MIT
url_site = "https://www.mit.edu/"
print(f"Navegando para {url_site}...")
driver.get(url_site)

# espera dois segundos para a página carregar
time.sleep(2)

# pega o título da página para verificar se foi carregada
titulo_da_pagina = driver.title
print(f"Página carregada - Título: '{titulo_da_pagina}'")
```

## Testes de componentes

```python
"""
#01 - Testa a estrutura principal da navbar. Verifica se:
1. O container do logo (site-nav__logo-link) existe e está visível
2. O container de links (site-nav__links) existe e está visível
3. O container de botões auxiliares (site-nav__buttons) existe e está visível
"""
try:
    print("Iniciando teste do item: Estrutura da Barra de Navegação...")
    
    # busca o container do logo
    seletor_logo = (By.CSS_SELECTOR, ".site-nav__logo-link")
    nav_logo = wait.until(EC.presence_of_element_located(seletor_logo))
    
    # 1) verifica se o logo está visível
    assert nav_logo.is_displayed(), "Falha: O container do logo não está visível."
    print("1) Container do logo (site-nav__logo-link) existe e está visível.")
    
    # busca o container de links principais
    seletor_links = (By.CSS_SELECTOR, ".site-nav__links")
    nav_links = wait.until(EC.presence_of_element_located(seletor_links))
    
    # 2) verifica se a lista de links está visível
    assert nav_links.is_displayed(), "Falha: O container de links não está visível."
    print("2) Container de links (site-nav__links) existe e está visível.")
    
    # busca o container de botões (pesquisa/menu mobile)
    seletor_botoes = (By.CSS_SELECTOR, ".site-nav__buttons")
    nav_botoes = wait.until(EC.presence_of_element_located(seletor_botoes))
    
    # 3) verifica se os botões estão visíveis
    assert nav_botoes.is_displayed(), "Falha: O container de botões não está visível."
    print("3) Container de botões (site-nav__buttons) existe e está visível.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

## Testes de imagem

```python
"""
#01 - Acha o svg do logo da MIT. Verifica se:
1. O Logo existe no DOM
2. Está visível na página
"""
try:
    print("Procura logo da faculdade...")
    
    # busca anchor tag que contém o logo
    seletor_logo = (By.CSS_SELECTOR, "a.site-nav__logo-link")
    
    # espera o logo ficar visível na tela
    logo = wait.until(EC.visibility_of_element_located(seletor_logo))
    
    # 1) verifica se o elemento existe no DOM
    assert logo is not None, "Falha: O elemento do logo não foi encontrado no DOM."
    print("1) Logo da MIT existe no DOM")
    
    # 2) verifica se o logo está sendo exibido
    assert logo.is_displayed(), "Falha: O logo existe no código, mas está oculto na tela."
    print("2) Logo está visível na página")
    
    # verifica dimensões do logo (site deixou o atributo de largura especificado no HTML)
    svg_interno = logo.find_element(By.XPATH, ".//*[local-name()='svg']")
    largura_logo = svg_interno.get_attribute("width")
    
except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#02 - Testa a imagem de destaque (Spotlight). Verifica se:
1. A imagem existe no DOM pela classe CSS e está visível na tela.
2. O texto alternativo (alt) está correto para acessibilidade.
3. O caminho da imagem (src) corresponde ao esperado.
"""
try:
    print("Procura Imagem de Destaque (Spotlight)...")
    
    # busca a imagem usando a classe de CSS (importante pois essa imagem pode mudar com o tempo)
    seletor_imagem = (By.CSS_SELECTOR, "img.spotlight__img")
    
    # espera a imagem ficar visível na tela
    imagem_destaque = wait.until(EC.visibility_of_element_located(seletor_imagem))
    
    # 1) verifica se o elemento existe no DOM e está visível
    assert imagem_destaque is not None, "Falha: A imagem de destaque não foi encontrada no DOM."
    assert imagem_destaque.is_displayed(), "Falha: A imagem existe no código, mas está oculta na tela."
    print("1) Imagem de destaque foi encontrada pela classe CSS e está visível.")
    
    # 2) verifica o texto alternativo (alt)
    texto_alt = imagem_destaque.get_attribute("alt")
    texto_esperado = "Sara Brown's hand reaches into a red-tinted stage model"
    assert texto_alt == texto_esperado, f"Falha: Texto alternativo incorreto. Encontrado: '{texto_alt}'"
    print("2) Texto alternativo (alt) configurado corretamente.")

    # 3) verifica se a imagem possui o src esperado
    src_imagem = imagem_destaque.get_attribute("src")
    assert "/files/images/202605/MIT-Sara-Brown-SL.jpg" in src_imagem, f"Falha: O caminho da imagem mudou. Atual: {src_imagem}"
    print("3) Caminho da fonte da imagem (src) validado com sucesso.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

## Testes de links

```python
"""
#02 - Testa o link do canal de youtube da MIT. Verifica se:
1. O link manda para o canal certo.
2. O texto interno com texto de acessibilidade existe e está correto.
"""

try:
    print("Procura link do canal de youtube no footer...")
    
    # buscando o link exato pelo seu destino (href)
    seletor_youtube = (By.CSS_SELECTOR, "a[href='https://www.youtube.com/mit']")
    
    # aguarda o elemento estar presente no DOM
    link_youtube = wait.until(EC.presence_of_element_located(seletor_youtube))
    
    # 1) verifica o destino do link
    destino = link_youtube.get_attribute("href")
    assert destino == "https://www.youtube.com/mit", f"Falha: URL de destino incorreta. Encontrado: {destino}"
    print("1) O link aponta para o canal correto do MIT no YouTube.")

    # 2) Busca a tag span com a classe visually-hidden relativa ao link para validação de acessibilidade (screen reader)
    span_acessibilidade = link_youtube.find_element(By.XPATH, ".//span[@class='visually-hidden']")
    texto_acessibilidade = span_acessibilidade.get_attribute("innerHTML").strip()
    assert texto_acessibilidade == "MIT@youtube", f"Falha: Texto de acessibilidade incorreto. Encontrado: {texto_acessibilidade}"
    print("2) Texto para leitores de tela configurado corretamente.")
    
except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#03 - Testa o link do vídeo 'A world without MIT'. Verifica se:
1. O vídeo abre em outra aba
2. A imagem do link existe e esá visível
"""
try:
    print("Procura bloco de vídeo incorporado...")
    
    # busca a anchor tag com base no seu link
    seletor_link = (By.CSS_SELECTOR, "a[href='https://www.youtube.com/watch?v=04JCB9GT64c']")
    # espera a anchor tag carregar no DOM
    link_video = wait.until(EC.presence_of_element_located(seletor_link))
    
    # verifica se o vídeo abre em outra aba
    target_attr = link_video.get_attribute("target")
    assert target_attr == "_blank", f"Falha: O link não abre em nova aba. Target: {target_attr}"
    print("1) Link configurado corretamente para abrir em nova aba (target='_blank').")

    # usa XPath para encontrar a tag de imagem dentro da anchro tag do vídeo encontrada
    imagem = link_video.find_element(By.XPATH, ".//img")
    
    atributo_alt = imagem.get_attribute("alt")
    atributo_loading = imagem.get_attribute("loading")
    
    assert atributo_alt == "Killian columns", f"Falha: Alt text incorreto. Encontrado: '{atributo_alt}'"
    assert atributo_loading == "lazy", f"Falha: Imagem sem lazy load. Encontrado: '{atributo_loading}'"
    print("2) Imagem possui texto alternativo correto e atributo loading='lazy'.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```
