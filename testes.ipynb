## Configuração da biblioteca

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
# define chrome como navegador
chrome_options = Options()

# executa chrome
driver = webdriver.Chrome(options=chrome_options)

# configura o webdriver para esperar no máximo dez segundos
wait = WebDriverWait(driver, 10)
```

```python
# vai para a página da MIT
url_site = "https://www.mit.edu/"
print(f"Navegando para {url_site}...")
driver.get(url_site)

# espera dois segundos para a página carregar
time.sleep(2)

# pega o título da página para verificar se foi carregada
titulo_da_pagina = driver.title
print(f"Página carregada - Título: '{titulo_da_pagina}'")
```

## Testes de componentes

```python
"""
#01 - Testa a estrutura principal da navbar. Verifica se:
1. O container do logo (site-nav__logo-link) existe e está visível
2. O container de links (site-nav__links) existe e está visível
3. O container de botões auxiliares (site-nav__buttons) existe e está visível
"""
try:
    print("Iniciando teste do item: Estrutura da Barra de Navegação...")
    
    # busca o container do logo
    seletor_logo = (By.CSS_SELECTOR, ".site-nav__logo-link")
    nav_logo = wait.until(EC.presence_of_element_located(seletor_logo))
    
    # 1) verifica se o logo está visível
    assert nav_logo.is_displayed(), "Falha: O container do logo não está visível."
    print("1) Container do logo (site-nav__logo-link) existe e está visível.")
    
    # busca o container de links principais
    seletor_links = (By.CSS_SELECTOR, ".site-nav__links")
    nav_links = wait.until(EC.presence_of_element_located(seletor_links))
    
    # 2) verifica se a lista de links está visível
    assert nav_links.is_displayed(), "Falha: O container de links não está visível."
    print("2) Container de links (site-nav__links) existe e está visível.")
    
    # busca o container de botões (pesquisa/menu mobile)
    seletor_botoes = (By.CSS_SELECTOR, ".site-nav__buttons")
    nav_botoes = wait.until(EC.presence_of_element_located(seletor_botoes))
    
    # 3) verifica se os botões estão visíveis
    assert nav_botoes.is_displayed(), "Falha: O container de botões não está visível."
    print("3) Container de botões (site-nav__buttons) existe e está visível.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#03 - Testa a imagem de destaque (Spotlight). Verifica se:
1. A imagem existe no DOM pela classe CSS e está visível na tela.
2. O texto alternativo (alt) está correto para acessibilidade.
"""
try:
    print("Procura Imagem de Destaque (Spotlight)...")
    
    # busca a imagem usando a classe de CSS (importante pois essa imagem pode mudar com o tempo)
    seletor_imagem = (By.CSS_SELECTOR, "img.spotlight__img")
    
    # espera a imagem ficar visível na tela
    imagem_destaque = wait.until(EC.visibility_of_element_located(seletor_imagem))
    
    # 1) verifica se o elemento existe no DOM e está visível
    assert imagem_destaque is not None, "Falha: A imagem de destaque não foi encontrada no DOM."
    assert imagem_destaque.is_displayed(), "Falha: A imagem existe no código, mas está oculta na tela."
    print("1) Imagem de destaque foi encontrada pela classe CSS e está visível.")
    
    # 2) verifica o texto alternativo (alt) (VAI FALHAR)
    texto_alt = imagem_destaque.get_attribute("alt")
    texto_esperado = "Sara Brown's hand reaches into a red-tinted stage model"
    assert texto_alt == texto_esperado, f"Falha: Texto alternativo incorreto. Encontrado: '{texto_alt}'"
    print("2) Texto alternativo (alt) configurado corretamente.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#04 - Testa o campo de input da busca. Verifica se:
1. O campo de input existe no DOM e está visível
2. É possível inserir texto e o valor do campo é atualizado
3. É possível limpar o texto inserido
4. O botão de fechar funciona e a tela retorna ao estado anterior
"""
try:
    print("Iniciando teste do item: Campo de Input da Busca...")
    
    # busca o input pelo seu ID
    seletor_input = (By.ID, "es-search-form-input")
    
    # espera o input ficar visível e pronto para interação
    input_busca = wait.until(EC.visibility_of_element_located(seletor_input))
    
    # 1) verifica se o input está visível
    assert input_busca.is_displayed(), "Falha: O input de busca não está visível na tela."
    print("1) Campo de input existe e está visível.")
    
    # 2) interage com o campo simulando a digitação de um usuário
    texto_teste = "selenium"
    input_busca.send_keys(texto_teste)
    # pega a propriedade 'value' para ver se o texto foi realmente digitado
    valor_digitado = input_busca.get_attribute("value")
    assert valor_digitado == texto_teste, f"Falha: Texto não foi inserido corretamente. Encontrado: '{valor_digitado}'"
    print("2) Inserção de texto validada: o valor mudou de acordo com a digitação.")
    
    # 3) testa a função de limpar o campo
    input_busca.clear()
    valor_limpo = input_busca.get_attribute("value")
    assert valor_limpo == "", f"Falha: O campo não foi limpo corretamente. Encontrado: '{valor_limpo}'"
    print("3) Limpeza do campo validada: o input voltou a ficar vazio.")
    
    # clica no botão de reset para voltar para a tela anterior
    seletor_botao_fechar = (By.CSS_SELECTOR, "button.js-search-reset")
    botao_fechar = wait.until(EC.element_to_be_clickable(seletor_botao_fechar))
    botao_fechar.click()
    time.sleep(1)

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

## Testes de links

```python
"""
#05 - Testa o link do canal de youtube da MIT. Verifica se:
1. O link manda para o canal certo.
2. O texto interno com texto de acessibilidade existe e está correto.
"""

try:
    print("Procura link do canal de youtube no footer...")
    
    # buscando o link exato pelo seu destino (href)
    seletor_youtube = (By.CSS_SELECTOR, "a[href='https://www.youtube.com/mit']")
    
    # aguarda o elemento estar presente no DOM
    link_youtube = wait.until(EC.presence_of_element_located(seletor_youtube))
    
    # 1) verifica o destino do link
    destino = link_youtube.get_attribute("href")
    assert destino == "https://www.youtube.com/mit", f"Falha: URL de destino incorreta. Encontrado: {destino}"
    print("1) O link aponta para o canal correto do MIT no YouTube.")

    # 2) Busca a tag span com a classe visually-hidden relativa ao link para validação de acessibilidade (screen reader)
    span_acessibilidade = link_youtube.find_element(By.XPATH, ".//span[@class='visually-hidden']")
    texto_acessibilidade = span_acessibilidade.get_attribute("innerHTML").strip()
    assert texto_acessibilidade == "MIT@youtube", f"Falha: Texto de acessibilidade incorreto. Encontrado: {texto_acessibilidade}"
    print("2) Texto para leitores de tela configurado corretamente.")
    
except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#06 - Testa o link do vídeo 'A world without MIT'. Verifica se:
1. O vídeo abre em outra aba
2. A imagem do link existe e esá visível
"""
try:
    print("Procura bloco de vídeo incorporado...")
    
    # busca a anchor tag com base no seu link
    seletor_link = (By.CSS_SELECTOR, "a[href='https://www.youtube.com/watch?v=04JCB9GT64c']")
    # espera a anchor tag carregar no DOM
    link_video = wait.until(EC.presence_of_element_located(seletor_link))
    
    # verifica se o vídeo abre em outra aba
    target_attr = link_video.get_attribute("target")
    assert target_attr == "_blank", f"Falha: O link não abre em nova aba. Target: {target_attr}"
    print("1) Link configurado corretamente para abrir em nova aba (target='_blank').")

    # usa XPath para encontrar a tag de imagem dentro da anchro tag do vídeo encontrada
    imagem = link_video.find_element(By.XPATH, ".//img")
    
    atributo_alt = imagem.get_attribute("alt")
    atributo_loading = imagem.get_attribute("loading")
    
    assert atributo_alt == "Killian columns", f"Falha: Alt text incorreto. Encontrado: '{atributo_alt}'"
    assert atributo_loading == "lazy", f"Falha: Imagem sem lazy load. Encontrado: '{atributo_loading}'"
    print("2) Imagem possui texto alternativo correto e atributo loading='lazy'.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#07 - Testa o link de endereço no rodapé. Verifica se:
1. O link com o endereço da universidade existe e está visível no rodapé.
2. O atributo href contém a URL do Google Maps.
3. O clique redireciona a janela do navegador para o Google Maps.
"""
try:
    print("Iniciando teste do item: Link de Endereço (Google Maps)...")
    
    # Força o scroll até o final absoluto da página para exibir o rodapé
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(1) # Aguarda a rolagem estabilizar
    
    # Busca o link do endereço usando a classe CSS fornecida
    seletor_endereco = (By.CSS_SELECTOR, "a.site-footer__contact-info-link")
    link_endereco = wait.until(EC.visibility_of_element_located(seletor_endereco))
    
    # 1) Verifica se o link está visível e checa o texto interno
    assert link_endereco.is_displayed(), "Falha: O link de endereço não está visível."
    texto_esperado = "77 Massachusetts Avenue, Cambridge, MA, USA"
    texto_link = link_endereco.get_attribute("innerHTML").strip()
    assert texto_esperado in texto_link, f"Falha: Texto do endereço incorreto. Encontrado: '{texto_link}'"
    print("1) Link de endereço encontrado e texto validado com sucesso.")
    
    # 2) Verifica se o atributo href realmente contém a URL de mapas do Google
    destino_href = link_endereco.get_attribute("href")
    assert "www.google.com/maps/" in destino_href, f"Falha: O link não aponta para o Google Maps. Encontrado: {destino_href}"
    print("2) Atributo href validado apontando para o serviço do Google Maps.")
    
    # 3) Clica no link via JavaScript para acionar a navegação
    driver.execute_script("arguments[0].click();", link_endereco)
    
    # Aguarda o redirecionamento. O Google Maps costuma ter "maps" na URL após o redirecionamento final
    wait.until(EC.url_contains("maps"))
    print("3) Clique efetuado e redirecionamento para o ambiente do Google Maps validado com sucesso.")
    
    # Retorna o navegador para a página anterior (MIT) para manter o estado do teste limpo
    driver.back()
    print("- Navegador retornou à página do MIT com sucesso.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

## Testes de responsividade

```python
"""
#09 - Testa o botão de navegação mobile. Verifica se:
1. O botão 'Menu' fica visível na resolução de celular.
2. Os elementos internos estão corretos.
3. O botão é clicável (usando JS para evitar interceptação).
"""
try:
    print("Iniciando teste do item: Botão de Menu Mobile...")
    
    # Redimensiona para celular
    driver.set_window_size(414, 896)
    
    # Fecha qualquer overlay de busca que possa estar aberto de testes anteriores
    try:
        botao_reset = driver.find_element(By.CSS_SELECTOR, "button.js-search-reset")
        if botao_reset.is_displayed():
            driver.execute_script("arguments[0].click();", botao_reset)
    except:
        pass
    
    # Garante scroll para o topo
    driver.execute_script("window.scrollTo(0, 0);")
    time.sleep(1)
    
    # Busca o botão mobile
    seletor_botao_mobile = (By.CSS_SELECTOR, "button.js-mobile-nav-trigger")
    botao_mobile = wait.until(EC.visibility_of_element_located(seletor_botao_mobile))
    
    # 1) verifica se o botão está visível
    assert botao_mobile.is_displayed(), "Falha: O botão de menu mobile não está visível."
    print("1) Botão 'Menu' visível na resolução de celular.")
    
    # 2) verifica elementos internos
    span_texto = botao_mobile.find_element(By.XPATH, ".//span[contains(@class, 'js-mobile-nav-trigger-text')]")
    texto_interno = span_texto.get_attribute("innerHTML").strip()
    assert texto_interno == "Menu", f"Falha: Texto do botão incorreto. Encontrado: '{texto_interno}'"
    print("2) Texto interno do botão validado ('Menu').")
    
    # 3) clica no botão via JS para evitar 'element click intercepted'
    driver.execute_script("arguments[0].click();", botao_mobile)
    print("3) Botão clicado com sucesso (via JS).")
    
    # Restaura o tamanho para desktop
    driver.maximize_window()

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    print(f"Erro Inesperado: {e}")
```

```python
"""
#10 - Testa o botão 'More' e seu menu dropdown em telas médias (tablet). Verifica se:
1. O botão 'More' fica visível ao forçar uma resolução de tela maior.
2. O botão é clicável e a ação de clique funciona.
3. O menu dropdown contendo os links auxiliares se torna visível após o clique.
4. O dropdown contém opções de navegação dentro dele.
"""

try:
    print("Iniciando teste do item: Botão 'More' (Dropdown Responsivo)...")
    
    # Redimensiona a janela para simular um tablet
    driver.set_window_size(1024, 768)
    # Recarrega a página para os eventos funcionarem corretamente (botão não estava funcionando sem isso)
    driver.refresh()
    # Força o scroll para o topo absoluto para garantir que a barra não esteja oculta
    driver.execute_script("window.scrollTo(0, 0);")
    # Busca o botão 'More'
    seletor_botao_more = (By.XPATH, "//button[contains(., 'More') and contains(@class, 'site-nav__link')]")
    # Espera o botão estar clicável
    botao_more = wait.until(EC.element_to_be_clickable(seletor_botao_more))
    
    # 1) verifica se o botão existe e está visível
    assert botao_more.is_displayed(), "Falha: O botão 'More' não está visível nesta resolução (1024x768)."
    print("1) Botão 'More' visível na resolução atual.")
    
    # 2) clica no botão 
    botao_more.click()
    print("2) Botão 'More' clicado com sucesso.")
    # Busca a lista ul que contém os sub-itens do menu
    seletor_dropdown = (By.CSS_SELECTOR, "ul.site-nav__more-list")
    # Espera a animação do dropdown terminar e ele ficar visível
    dropdown = wait.until(EC.visibility_of_element_located(seletor_dropdown))
    
    # 3) verifica se o menu dropdown realmente abriu
    assert dropdown.is_displayed(), "Falha: O menu dropdown não apareceu na tela após clicar no botão."
    print("3) Menu dropdown expandiu e está visível.")
    
    # 4) verifica se existem opções dentro do dropdown
    opcoes = dropdown.find_elements(By.TAG_NAME, "li")
    assert len(opcoes) > 0, "Falha: O dropdown abriu, mas está vazio (sem opções de link)."
    print(f"4) Dropdown validado com sucesso contendo {len(opcoes)} opções de navegação.")

    # Restaura o tamanho da janela para desktop
    driver.maximize_window()

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#12 - Testa a navegação para a página 'About' e o vídeo em destaque. Verifica se:
1. O link 'About MIT' é clicável e redireciona a janela para a rota correta.
2. O vídeo da página 'About' existe no DOM e fica visível na tela.
3. Os atributos de controle do vídeo (autoplay, loop, muted) estão ativos.
4. A tag <source> aponta para o caminho correto do arquivo .mp4.
"""
try:
    print("Iniciando teste do item: Navegação 'About MIT' e Validação de Vídeo...")
    
    # Redimensiona para desktop para garantir que o menu superior está visível
    driver.maximize_window()
    # Garante que o teste comece na página inicial (Home)
    driver.get("https://web.mit.edu/")
    # Busca o link 'About MIT' no cabeçalho
    seletor_link_about = (By.XPATH, "//a[contains(@class, 'site-nav__link') and text()='About MIT']")
    link_about = wait.until(EC.element_to_be_clickable(seletor_link_about))
    # Clica no link
    link_about.click()
    
    # 1) Aguarda até que a URL do navegador contenha a rota '/about'
    wait.until(EC.url_contains("/about"))
    print("1) Link 'About MIT' clicado e redirecionamento para a nova página validado.")
    
    # Busca a tag <video> pela classe de CSS
    seletor_video = (By.CSS_SELECTOR, "video.pause-video")
    video_about = wait.until(EC.visibility_of_element_located(seletor_video))
    # 2) Verifica se o vídeo carregou e está sendo exibido
    assert video_about.is_displayed(), "Falha: O vídeo da página About não está visível na tela."
    print("2) Elemento de vídeo encontrado e renderizado com sucesso na página 'About'.")
    
    # 3) Verifica os atributos de reprodução automática do HTML5
    attr_autoplay = video_about.get_attribute("autoplay")
    attr_loop = video_about.get_attribute("loop")
    attr_muted = video_about.get_attribute("muted")
    # obs: atributos booleanos HTML (como loop, muted) retornam a string "true" se estiverem presentes em vez de um valor booleano
    assert attr_autoplay == "true", "Falha: O vídeo não está configurado para iniciar automaticamente (autoplay)."
    assert attr_loop == "true", "Falha: O vídeo não está em loop."
    assert attr_muted == "true", "Falha: O vídeo não está configurado como mudo (muted)."
    print("3) Atributos nativos de reprodução (autoplay, loop, muted) validados.")
    
    # 4) Verifica o caminho do arquivo de origem buscando a tag interna <source>
    tag_source = video_about.find_element(By.TAG_NAME, "source")
    src_arquivo = tag_source.get_attribute("src")
    
    assert "/files/images/201807/about-MIT-5.mp4" in src_arquivo, f"Falha: O caminho do arquivo .mp4 está incorreto. Encontrado: {src_arquivo}"
    print("4) Fonte de origem do vídeo (src) validada com sucesso.")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```

```python
"""
#17 - Testa o envio do formulário de pesquisa. Verifica se:
1. É possível digitar 'Tim Berners-Lee' e enviar a pesquisa com a tecla ENTER.
2. O navegador é redirecionado para a página de resultados com a URL correta.
"""

# necessário para testar input do teclado
from selenium.webdriver.common.keys import Keys

try:
    print("Iniciando teste do item: Realizar Pesquisa (Tim Berners-Lee)...")
    
    # vai para o topo da página e deixa a largura máxima
    driver.maximize_window()
    driver.execute_script("window.scrollTo(0, 0);")
    
    # clica no ícone da lupa para abrir a barra de pesquisa
    seletor_lupa = (By.CSS_SELECTOR, "a.js-search-trigger")
    lupa = wait.until(EC.element_to_be_clickable(seletor_lupa))
    driver.execute_script("arguments[0].click();", lupa)
    # busca o input da barra de pesquisa
    seletor_input = (By.ID, "es-search-form-input")
    input_busca = wait.until(EC.element_to_be_clickable(seletor_input))
    
    # Clica na barra de input
    input_busca.click()
    # 2) digita o termo e pressiona ENTER para disparar o formulário
    termo = "Tim Berners-Lee"
    input_busca.send_keys(termo)
    # envia a tecla RETURN (Enter) para o input
    input_busca.send_keys(Keys.RETURN)
    print(f"2) Termo '{termo}' inserido e tecla ENTER acionada.")
    
    # 2) aguarda o redirecionamento e verifica a URL
    wait.until(EC.url_contains("/search"))
    url_atual = driver.current_url
    assert "q=" in url_atual, f"Falha: A URL não possui o parâmetro de query de pesquisa. URL: {url_atual}"
    print("2) Redirecionamento validado com sucesso: O navegador acessou a página de resultados (/search).")

except AssertionError as erro_teste:
    print(f"Erro de Asserção: {erro_teste}")
except Exception as e:
    # erros inesperados (ex: timeout do selenium)
    print(f"Erro Inesperado durante a execução do teste: {e}")
```
